```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../data/raw/nvda.csv", index_col="Date", parse_dates=True)
df = df.iloc[:-1]
df.tail()
```

```python
df["Close"].plot(figsize=(12, 5), title="NVDA - precio de cierre")
plt.show()
```

```python
df["retorno"] = df["Close"].pct_change()
vol_diaria = df["retorno"].std()
vol_anual = vol_diaria * (252 ** 0.5)
print(f"Volatilidad diaria: {vol_diaria:.2%}")
print(f"Volatilidad anualizada: {vol_anual:.2%}")
```

```python
df["drawdown"] = df["Close"] / df["Close"].cummax() - 1
print(f"Peor caída desde un máximo: {df['drawdown'].min():.2%}")
df["drawdown"].plot(figsize=(12, 4), title="Drawdown de NVDA")
plt.show()
```
